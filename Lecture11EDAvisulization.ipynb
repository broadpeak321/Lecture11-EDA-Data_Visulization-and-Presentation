```python
#product sale ratio
product = ['tshirt', 'casual-shirt', 'jeans', 'dress-shirt']   # (cat) Quantitaive nominal
sales = [23, 64, 12, 19]   #value   (qualitative discrete)
```

```python
plt.bar(product,sales)
plt.xlabel('Products')
plt.ylabel('Sale Quantity')
plt.title('Product Sales Comparison')
plt.show()
```

```python
plt.barh(product,sales, color='red', edgecolor='black', linewidth=1.2,)
plt.xlabel('Products')
plt.ylabel('Sale Quantity')
plt.title('Product Sales Comparison')
plt.show()
```

```python
# for studing the relation among, focus is on saturation
plt.pie(sales, labels=product)
plt.show()
```
